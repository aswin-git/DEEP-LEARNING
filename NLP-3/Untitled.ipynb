```python
from gensim.summarization import summarize

text = """
Machine learning is a field of artificial intelligence that focuses on teaching machines
how to learn from data. It is widely used in applications such as recommendation systems,
image recognition, speech processing, and natural language processing. Machine learning
models improve their performance as they are exposed to more data over time.
"""

summary_textrank = summarize(text, ratio=0.4)
print(summary_textrank)
```

```python
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer

text = """
Machine learning is a field of artificial intelligence that focuses on teaching machines
how to learn from data. It is widely used in applications such as recommendation systems,
image recognition, speech processing, and natural language processing. Machine learning
models improve their performance as they are exposed to more data over time.
"""

parser = PlaintextParser.from_string(text, Tokenizer("english"))
summarizer = TextRankSummarizer()

summary = summarizer(parser.document, sentences_count=2)

for sentence in summary:
    print(sentence)
```
